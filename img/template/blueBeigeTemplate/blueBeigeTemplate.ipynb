**Creating dataframes**

Data source: *Source: sample data*

```python
import numpy as np

np.random.seed(44)

length = 21

serie1 = np.random.randint(50, 100, length)
serie2 = np.random.randint(40, 150, length)
serie3 = np.random.randint(0, 60, length)

cathegory_data_multiseries = pd.DataFrame({
    "year": [y for y in range(1990, 1990 + length)],
    "serie1": serie1,
    "serie2": serie2,
    "serie3": serie3,
})
```

```python
cathegory_data_multiseries_long = cathegory_data_multiseries.melt(
    id_vars = "year",
    value_vars = ["serie1", "serie2", "serie3"],
    var_name = "serie",
    value_name = "values"
)
```

Data story with **line chart**

```python
story = (
    pn.Story(
    #Builing the Story class object
        data = cathegory_data_multiseries_long,
        width = 800,
        height = 400,
        template = blueBeigeTemplate #Using blueBeigeTemplate
    )

    #Method invocation
    #Line chart
    .mark_line(
        point = True,
        interpolate = "cardinal"
    )

    #Data encoding
    .encode(
        x = alt.X(
            "year:O",
            title = "", 
            axis = alt.Axis(
                labelAngle = -30
                ),
        ),
        y = alt.Y(
            "values:Q",
            title = "",
            axis = alt.Axis(grid = True)
        ),
        color = alt.Color("serie:N", title = "Series"),
    )
    
    #Data source
    .add_source(
        text = "Source: sample data",
        position = "top",
        align = "left"
    )
    
    #Title and subtitle
    .add_title(
        title = "Example of using pynarrative with templates",
        subtitle = "Example of using pynarrative with templates",
        align = "center"
    )


    .add_context(
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
        position = "top",
    )

    .add_next_steps(
        mode = "vertical",
        position = "right",
        steps = [
            "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam",
            "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam",
            "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam",
            "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam"
            ]
    )

    .render()
)

story
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)

n_samples = 100

sq_meters = np.random.uniform(40, 250, n_samples)

locations = np.random.choice(
    ["City Center", "Suburbs", "Rural"], size=n_samples, p=[0.3, 0.5, 0.2]
)

multiplier = {"City Center": 4500, "Suburbs": 2500, "Rural": 1200}
noise = np.random.normal(0, 25000, n_samples)

price_eur = [
    (m2 * multiplier[loc]) + n
    for m2, loc, n in zip(sq_meters, locations, noise)
]

sq_price_location = pd.DataFrame(
    {
        "sq_meters": np.round(sq_meters, 1),
        "price": np.round(price_eur, -2),
        "location": locations,
    }
)
```

```python
story = (
    pn.Story(
    #Builing the Story class object
        data = sq_price_location,
        width = 600,
        height = 300,
        template = blueBeigeTemplate #Using blueBeigeTemplate
    )

    #Method invocation
    #Bar chart
    .mark_point()

    #Data encoding
    .encode(
        x = alt.X(
            "sq_meters:Q",
            title = "Square feet", 
            axis = alt.Axis(grid = True),
        ),
        y = alt.Y(
            "price:Q",
            title = "Price (€)",
            axis = alt.Axis(grid = True)
        ),
        color = alt.Color("location:N", title = "Location"),
    )
    
    #Data source
    .add_source(
        text = "Source: sample data",
        position = "top",
        align = "left"
    )
    
    #Title and subtitle
    .add_title(
        title = "Example of using pynarrative with templates",
        subtitle = "Example of using pynarrative with templates",
        align = "center"
    )

    .add_context(
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt.",
        position = "top",
        font_weight = "bold"
    )

    .add_context(
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
        position = "bottom",
    )

    .add_next_steps(
        mode = "vertical",
        position = "right",
        steps = [
            "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam",
            "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam",
            ]
    )

    .render()
)

story
```

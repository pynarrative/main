```python
year = [y  for y in range(2012, 2025)]

year_values = pd.DataFrame({
    "year" : year,
    "value" : [5.201, 5.625, 6.182, 6.551, 6.409, 7.036, 7.650, 7.618, 1.086, 1.689, 9.812, 12.298, 14.733]
})

story = (
    pn.Story(
    #Builing the Story class object
        data = year_values,
        template = areaTemplate,
    )

    #Method invocation
    #Area chart
    .mark_area(
        interpolate = "cardinal",
        color = "rgb(242, 80, 80)"
    )

    #Data encoding
    .encode(
        #We need both quantitative axis to add annotation
        x = alt.X(
            "year:Q",
            title = "Year", 
            axis = alt.Axis(
                format = "d",
                labelAngle = -30,
                values = list(range(2012, 2025))
            )
        ),
        y = alt.Y(
            "value:Q",
            title = "Value (in millions)",
            scale = alt.Scale(domain=[0, 20])
        ),
    )
    
    #Data source
    .add_source(
        text = "Source: sample data",
        position = "top",
        align = "left"
    )
    
    #Title and subtitle
    .add_title(
        title = "Example of using pynarrative with areaTemplate",
        subtitle = "areaTemplate",
        align = "center"
    )

    .add_labels_chart(
        values = "value:Q",
        font_weight = "bold",
        font_size = 14,
        dy = -15
    )

    .add_annotation(
        x = 2015,
        y = 14.000,
        text = "Lorem ipsum dolor sit amet, consectetur adipiscing elit. Quisque euismod turpis eget nisl cursus venenatis.",
    )

    .add_line(
        orientation = "horizontal",
        value = year_values["value"],
        math = "mean",
    )

    # Context (left)
    .add_context(
        position = "left",
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat. Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat."
    )

    # Next steps (bottom)
    .add_next_steps(
        position = "bottom",
        mode = "horizontal",
        steps = [
            "Lorem ipsum dolor sit amet, consectetur adipiscing elit. Quisque euismod turpis eget nisl cursus venenatis. Duis gravida nulla magna, sit amet fermentum magna sodales id. Donec nec enim justo. Sed non sollicitudin ex. Ut varius neque ut tellus scelerisque luctus. Suspendisse sodales libero odio, quis auctor tortor tristique in. Mauris sed sodales leo, quis rhoncus erat. Sed convallis et mi nec vulputate. Aliquam vel turpis at erat maximus commodo sit amet at justo. Proin egestas massa sit amet porttitor convallis. Ut sed nisl quis nisi tempor sagittis vitae sit amet mi. Aliquam erat volutpat.",
            "Lorem ipsum dolor sit amet, consectetur adipiscing elit. Quisque euismod turpis eget nisl cursus venenatis. Duis gravida nulla magna, sit amet fermentum magna sodales id. Donec nec enim justo. Sed non sollicitudin ex. Ut varius neque ut tellus scelerisque luctus. Suspendisse sodales libero odio, quis auctor tortor tristique in. Mauris sed sodales leo, quis rhoncus erat. Sed convallis et mi nec vulputate. Aliquam vel turpis at erat maximus commodo sit amet at justo. Proin egestas massa sit amet porttitor convallis. Ut sed nisl quis nisi tempor sagittis vitae sit amet mi. Aliquam erat volutpat.",
            "Lorem ipsum dolor sit amet, consectetur adipiscing elit. Quisque euismod turpis eget nisl cursus venenatis. Duis gravida nulla magna, sit amet fermentum magna sodales id. Donec nec enim justo. Sed non sollicitudin ex. Ut varius neque ut tellus scelerisque luctus. Suspendisse sodales libero odio, quis auctor tortor tristique in. Mauris sed sodales leo, quis rhoncus erat. Sed convallis et mi nec vulputate. Aliquam vel turpis at erat maximus commodo sit amet at justo. Proin egestas massa sit amet porttitor convallis. Ut sed nisl quis nisi tempor sagittis vitae sit amet mi. Aliquam erat volutpat."
        ],
        title_dx = -25
    )

    .render()
)
story
```

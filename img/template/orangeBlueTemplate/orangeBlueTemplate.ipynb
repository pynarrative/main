**Creating dataframes**

Data source: *Source: Statista (statista.com)*

```python
country = [
    "United States", "Argentina", "Georgia", "India", "Russia", "Romania",
    "Chile", "Armenia", "Moldova", "Bulgaria", "Latvia", "Montenegro",
    "Bosnia and Herzegovir", "Croatia", "Azerbaijan", "Serbia", "Albania",
    "New Zealand", "Turkey", "Belarus", "Poland", "Hungary", "Greece",
    "South Korea", "North Macedonia", "Ukraine", "Portugal", "Luxembourg",
    "Lithuania", "Slovenia", "Czechia", "Slovakia", "Italy", "Canada",
    "France", "Australia", "Belgium", "Israel", "Finland", "Estonia",
    "Austria", "Netherlands", "Malta", "Germany", "Denmark", "Ireland",
    "Spain", "Japan", "Iceland", "United Kingdom", "Switzerland", "Mexico",
    "Sweden", "Norway"
]

death = [
    133.98, 121.23, 118.06, 113.51, 100.58, 97.58, 94.94, 89.19, 88.60, 
    79.31, 77.32, 77.18, 73.52, 66.58, 63.32, 63.20, 61.45, 59.26, 59.13, 
    58.54, 55.88, 54.52, 54.45, 53.99, 53.04, 52.18, 51.26, 51.25, 51.23, 
    50.09, 46.29, 45.04, 44.60, 42.22, 41.55, 41.20, 41.02, 37.88, 37.61, 
    37.26, 36.59, 33.08, 31.79, 30.79, 25.36, 25.15, 23.99, 23.63, 23.22, 
    22.94, 20.90, 19.03, 18.42, 14.28
]

#Let's take the first 15 records
death_road_accidents = pd.DataFrame({
    "country" : country[:15],
    "death" : death[:15]
})
```

Data story with **bar chart**

```python
story = (
    pn.Story(
    #Builing the Story class object
        data = death_road_accidents, #Using the long (melted) dataframe!
        template = orangeBlueTemplate #Using orangeBlueTemplate
    )

    #Method invocation
    #Bar chart
    .mark_bar(
        cornerRadiusEnd = 7,
        size = 20
    )

    #Data encoding
    #To create a horizontal bar chart we plot the cathegorical data on Y and the numerical data on X
    .encode(
        y = alt.Y(
            "country:N",
            title = "Country", 
            axis = alt.Axis(
                grid = True,
            ),
            sort = "-x" #Sorting the data in a descending order
        ),
        x = alt.X(
            "death:Q",
            title = "Death in road accidents per million inhabitants",
            axis = alt.Axis(
                grid = True
            ),
        ),
    )
    
    #Data source
    .add_source(
        text = "Source: Statista (statista.com)",
        position = "left",
        align = "center"
    )
    
    #Title and subtitle
    .add_title(
        title = "Number of deaths due to road accidents per 1 million",
        subtitle = "by country, in 2024",
        align = "left",
        img_url = "https://cdn-icons-png.flaticon.com/512/65/65788.png",
        x_img_offset = 575,
        y_img_offset = -10,
    )

    #Context area (on the right)
    .add_context(
        text = "According to 2024 Statista data, road traffic fatalities reveal stark regional disparities. While countries like Azerbaijan (63.32) and Croatia (66.58) report lower relative mortality, seven nations exceed the average threshold of 93 deaths per million inhabitants—with the United States (133.98), Argentina (121.23), and Georgia (118.06) representing the highest rates.",
        position = "right",
    )

    #Highlighting line (mean)
    .add_line(
        orientation = "vertical",
        value = death_road_accidents["death"],
        math = "mean",
        label_text = "Average death in road accidents: 93 people per 1 million",
        label_font_size = 13,
        label_dx = -150,
    )

    .add_labels_chart(
        values = "death:Q",
        font_weight = "bold",
        font_size = 14,
        dx = 25
    )

    .render()
)

story
```

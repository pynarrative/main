**Creating dataframe**

Data source: *Société d'exploitation de la tour Eiffel via statista.com*

```python
year = [y  for y in range(2011, 2026)]

tour_eiffel = pd.DataFrame({
    "year" : year,
    "visitors" : [7.08, 6.27, 6.74, 7.10, 6.92, 5.84, 6.23, 6.07, 6.14, 1.16, 2.06, 5.85, 6.32, 6.30, 6.74]
})
```

Data story with **bar chart**

```python
story = (
    pn.Story(
    #Builing the Story class object
        data = tour_eiffel,
        width = 600,
        height = 300,
        template = tourEiffelTemplate #Using the tourEiffelTemplate
    )

    #Method invocation
    #Bar chart
    .mark_bar(
        cornerRadiusEnd = 10,
        size = 25
    )

    #Data encoding
    .encode(
        x = alt.X(
            "year:N",
            title = "Year", 
            axis = alt.Axis(
                grid = True,
                labelAngle = -45
            )
        ),
        y = alt.Y(
            "visitors:Q",
            title = "Number of visitors (Millions)",
            axis = alt.Axis(
                grid = True
            ),
            scale = alt.Scale(
                domain=[0, 16]
            )
        )
    )

    #Data source
    .add_source(
        text = "Source: Société d'exploitation de la tour Eiffel via statista.com",
        position = "top",
        align = "right"
    )
    
    #Title and subtitle
    .add_title(
        title = "Number of visitors to the Eiffel Tower in Paris",
        subtitle = "from 2011 to 2025",
        align = "center"
    )

    #Context area (on the left)
    .add_context(
        text = " The number of visitors to the Eiffel Tower in Paris, France, increased in 2025 over the previous year, totaling 6.74 million. Built between 1887 and 1889 as part of the preparations for the World's Fair, the Eiffel Tower is one of the most iconic symbols of Paris and France.",
        position = "left",
    )

    #Labels
    .add_labels_chart(
        values = "visitors:Q",
        font_size = 14,
        font_weight = "bold",
        dy = -10
    )

    #Mean
    .add_line(
        value = tour_eiffel["visitors"],
        math = "mean",
        label_font_size = 14,
    )

    #Highlighing line
    .add_line(
        orientation = "vertical",
        value = 2020,
        label_text = "Covid 19 pandemic beginning",
        color = "darkblue",
        label_dx = -200,
        label_dy = 50,
        label_font_size = 14
    )

    #Highlighing line
    .add_line(
        orientation = "vertical",
        value = 2021,
        label_text = "End of Covid 19 pandemic restricrions",
        color = "darkblue",
        label_dx = 5,
        label_dy = 75,
        label_font_size = 14
    )

    #Highlighing bar
    .add_highlight(2014)

    .render()
)

story
```

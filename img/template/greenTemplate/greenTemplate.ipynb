**Creating dataframes**

Data source: *Source: sample data*

```python
year_value = pd.DataFrame({
    "year": [i for i in range(2000, 2007)],
    "value": [0, 3, 5, 6, 6.2, 6.7, 8]
})
```

Data story with **line chart**

```python
story = (
    pn.Story(
    #Builing the Story class object
        data = year_value,
        template = greenTemplate #Using greenTemplate
    )

    #Method invocation
    #Line chart
    .mark_line(
        point = True,
        interpolate = "cardinal"
    )

    #Data encoding
    .encode(
        #We need both quantitative axis to add annotation
        x = alt.X(
            "year:Q",
            title = "", 
            axis = alt.Axis(
                format = "d",
                values = list(range(2000, 2007))
            ),
            # scale = alt.Scale(domain=[2000, 2006])
        ),
        y = alt.Y(
            "value:Q",
            title = "Value",
            scale = alt.Scale(domain=[0, 10])
        )
    )
    
    #Data source
    .add_source(
        text = "Source: sample data",
        position = "bottom",
        align = "left"
    )
    
    #Title and subtitle
    .add_title(
        title = "Example of using pynarrative with templates",
        align = "left"
    )

    .add_context(
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
        position = "top",
        title = ""
    )

    .add_labels_chart(
        values = "value:Q",
        font_weight = "bold",
        font_size = 14,
        dy = -15
    )

    .add_annotation(
        x = 2003,
        y = 6,
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat.",
        text_color = "#05053D",
        show_subject = True,
        dy = 41,
        dx = -40,
        background_color = "transparent"
    )

    .render()
)

story
```

Data story with **bar chart**

```python
story = (
    pn.Story(
    #Builing the Story class object
        data = year_value,
        template = greenTemplate #Using greenTemplate
    )

    #Method invocation
    #Bar chart
    .mark_bar(
        size = 40
    )

    #Data encoding
    .encode(
        #We need both quantitative axis to add annotation
        x = alt.X(
            "year:Q",
            title = "", 
            axis = alt.Axis(
                format = "d",
                values = list(range(2000, 2007))
            ),
            # scale = alt.Scale(domain=[2000, 2006])
        ),
        y = alt.Y(
            "value:Q",
            title = "Value",
            scale = alt.Scale(domain=[0, 12])
        )
    )
    
    #Data source
    .add_source(
        text = "Source: sample data",
        position = "bottom",
        align = "left"
    )
    
    #Title and subtitle
    .add_title(
        title = "Example of using pynarrative with templates",
        align = "left"
    )

    .add_context(
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
        position = "top",
        title = ""
    )

    .add_labels_chart(
        values = "value:Q",
        font_weight = "bold",
        font_size = 14,
        dy = -15
    )

    .add_annotation(
        x = 2003,
        y = 5,
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat.",
        text_color = "#05053D",
        dy = -250,
        dx = -170
    )

    .render()
)

story
```

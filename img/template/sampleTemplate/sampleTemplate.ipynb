**Creating dataframes**

Data source: *Source: sample data*

```python
import numpy as np
import string

np.random.seed(43)

len = 10
value = np.random.randint(0, 100, len)

data_value = pd.DataFrame({
    "cathegory": [i for i in string.ascii_uppercase[:len]],
    "value": value
})
```

Data story with **horizontal bar chart**

```python
story = (
    pn.Story(
    #Builing the Story class object
        data = data_value,
        width = 700,
        height = 400,
        template = sampleTemplate #Using sampleTemplate
    )

    #Method invocation
    #Bar chart
    .mark_bar()

    #Data encoding
    #To create a horizontal bar chart we plot the cathegorical data on Y and the numerical data on X
    .encode(
        y = alt.Y(
            "cathegory:N",
            title = "", 
            sort = "-x" #Sorting data in descending order
        ),
        x = alt.X(
            "value:Q",
            title = "",
        ),
    )
    
    #Data source
    .add_source(
        text = "Source: sample data",
        position = "bottom",
        align = "left",
    )
    
    #Title and subtitle
    .add_title(
        title = "Example of using pynarrative with templates",
        align = "left",
    )

    .add_context(
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
        position = "right",
        align = "top",
        title_dy = 20,
    )

    #Next steps area (on the right)
    .add_next_steps(
        steps = [
            "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
            ],
        position = "right",
        mode = "vertical",
        title_dy = -25
    )

    .add_highlight(91)

    .add_labels_chart(
        values = "value:Q",
        font_weight = "bold",
        font_size = 16,
        dx = -25,
        color = "white"
    )

    .render()
)

story
```

Data story with **bar chart**

```python
story = (
    pn.Story(
    #Builing the Story class object
        data = data_value,
        width = 750,
        height = 400,
        template = sampleTemplate #Using sampleTemplate
    )

    #Method invocation
    #Bar chart
    .mark_bar()

    #Data encoding
    .encode(
        x = alt.X(
            "cathegory:N",
            title = "", 
            sort = "-y" #Sorting data in descending order
        ),
        y = alt.Y(
            "value:Q",
            title = "",
        ),
    )
    
    #Data source
    .add_source(
        text = "Source: sample data",
        position = "bottom",
        align = "left"
    )
    
    #Title and subtitle
    .add_title(
        title = "Example of using pynarrative with templates",
        align = "left",
    )

    .add_context(
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
        position = "right",
        align = "top",
        title_dy = 20
    )

    #Next steps area (on the right)
    .add_next_steps(
        steps = [
            "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
            ],
        position = "right",
        mode = "vertical",
        text_align = "left",
        title_dy = -25
    )


    .add_highlight("J")

    .add_labels_chart(
        values = "value:Q",
        font_weight = "bold",
        font_size = 16,
        dy = 25,
        color = "white"
    )

    .render()
)

story
```
